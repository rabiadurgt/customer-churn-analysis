```python
from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_DIR = PROJECT_ROOT / "veri"
CLEAN_DIR = PROJECT_ROOT / "ciktilar" / "temiz_veriler"
OUTPUT_DIR = PROJECT_ROOT / "ciktilar"

vaka_metni = (DATA_DIR / "VAKA.md").read_text(encoding="utf-8")
veri_sozlugu_metni = (DATA_DIR / "veri_sozlugu.md").read_text(encoding="utf-8")

print("Proje klasörü:", PROJECT_ROOT)
print("\nVAKA.md ve veri_sozlugu.md başarıyla okundu.")

# Temiz veri setleri
musteriler = pd.read_csv(
    CLEAN_DIR / "musteriler_temiz.csv",
    encoding="utf-8-sig",
    parse_dates=["ilk_temas_tarihi"]
)

araclar = pd.read_csv(
    CLEAN_DIR / "araclar_temiz.csv",
    encoding="utf-8-sig",
    parse_dates=["satin_alma_tarihi", "garanti_bitis_tarihi"]
)

servis = pd.read_csv(
    CLEAN_DIR / "servis_kayitlari_temiz.csv",
    encoding="utf-8-sig",
    parse_dates=["servis_tarihi"]
)

geri_bildirimler = pd.read_csv(
    CLEAN_DIR / "geri_bildirimler_temiz.csv",
    encoding="utf-8-sig",
    parse_dates=["tarih"]
)

veri_setleri = {
    "Müşteriler": musteriler,
    "Araçlar": araclar,
    "Servis kayıtları": servis,
    "Geri bildirimler": geri_bildirimler
}

for ad, df in veri_setleri.items():
    print(f"{ad}: {df.shape[0]:,} satır, {df.shape[1]} sütun")
```

```python
tarih_sutunlari = {
    "Müşteriler": ["ilk_temas_tarihi"],
    "Araçlar": ["satin_alma_tarihi", "garanti_bitis_tarihi"],
    "Servis kayıtları": ["servis_tarihi"],
    "Geri bildirimler": ["tarih"]
}

for ad, sutunlar in tarih_sutunlari.items():
    df = veri_setleri[ad]

    print(f"\n--- {ad} ---")

    for sutun in sutunlar:
        tarihler = df[sutun]

        print(
            f"{sutun}: "
            f"min={tarihler.min()}, "
            f"max={tarihler.max()}, "
            f"NaT={tarihler.isna().sum()}, "
            f"tip={tarihler.dtype}"
        )
```

```python
# Analiz tarihleri
KESIM_TARIHI = pd.Timestamp("2025-06-30")
HEDEF_BASLANGIC = pd.Timestamp("2025-07-01")
HEDEF_BITIS = pd.Timestamp("2026-06-30")

# Kesim tarihine kadar olan servis kayıtları
servis_gozlem = servis[
    servis["servis_tarihi"] <= KESIM_TARIHI
].copy()

# Hedef dönemdeki servis kayıtları
servis_hedef = servis[
    (servis["servis_tarihi"] >= HEDEF_BASLANGIC) &
    (servis["servis_tarihi"] <= HEDEF_BITIS)
].copy()

# Gözlem döneminde en az bir servis kaydı bulunan araçlar
gozlem_araclari = servis_gozlem["arac_id"].nunique()

# Müşteri evreni: gözlem döneminde en az bir servis kaydı bulunan müşteriler
aktif_arac_musterileri = araclar[
    araclar["arac_id"].isin(servis_gozlem["arac_id"])
]["musteri_id"].unique()

analiz_musterileri = musteriler[
    musteriler["musteri_id"].isin(aktif_arac_musterileri)
].copy()

print("KESİM TARİHİ:", KESIM_TARIHI.date())
print("\n--- Gözlem dönemi ---")
print("Servis kayıt sayısı:", f"{len(servis_gozlem):,}")
print("Servis gören araç sayısı:", f"{gozlem_araclari:,}")
print("Analiz evrenindeki müşteri sayısı:", f"{len(analiz_musterileri):,}")

print("\n--- Hedef dönemi ---")
print("Servis kayıt sayısı:", f"{len(servis_hedef):,}")

print("\n--- Kontroller ---")
print("Müşteri ID tekil mi?", analiz_musterileri["musteri_id"].is_unique)
print("Gözlem servislerinde kesim tarihi sonrası kayıt var mı?",
      (servis_gozlem["servis_tarihi"] > KESIM_TARIHI).any())
print("Hedef servislerinde hedef başlangıcından önce kayıt var mı?",
      (servis_hedef["servis_tarihi"] < HEDEF_BASLANGIC).any())
```

```python
# Gözlem dönemi servis kayıtlarını müşteri ID'si ile eşleştirme
servis_musteri = servis_gozlem.merge(
    araclar[["arac_id", "musteri_id"]],
    on="arac_id",
    how="left",
    validate="many_to_one"
)

# Müşteri bazında RFM metrikleri
rfm = (
    servis_musteri
    .groupby("musteri_id")
    .agg(
        son_servis_tarihi=("servis_tarihi", "max"),
        frequency=("kayit_id", "count"),
        monetary=("tutar", "sum")
    )
    .reset_index()
)

# Recency: kesim tarihinden son servise kadar geçen gün
rfm["recency"] = (
    KESIM_TARIHI - rfm["son_servis_tarihi"]
).dt.days

# Kontroller
print("RFM tablosu satır sayısı:", f"{len(rfm):,}")
print("Tekil müşteri sayısı:", f"{rfm['musteri_id'].nunique():,}")

print("\nRFM özet istatistikleri:")
print(rfm[["recency", "frequency", "monetary"]].describe().round(2))

print("\nEksik değerler:")
print(rfm[["recency", "frequency", "monetary"]].isna().sum())

print("\nİlk 5 kayıt:")
display(rfm.head())
```

```python
# Monetary dağılımı ve tutar kontrolleri
print("--- Tutar dağılımı ---")
print(
    servis_musteri["tutar"]
    .describe(percentiles=[0.01, 0.05, 0.25, 0.50, 0.75, 0.90, 0.95, 0.99])
    .round(2)
)

print("\n--- Tutar kontrolü ---")
print("Negatif tutarlı servis kaydı:", (servis_musteri["tutar"] < 0).sum())
print("Sıfır tutarlı servis kaydı:", (servis_musteri["tutar"] == 0).sum())
print("Pozitif tutarlı servis kaydı:", (servis_musteri["tutar"] > 0).sum())

# En yüksek tutarlı 10 servis kaydı
print("\n--- En yüksek 10 servis tutarı ---")
display(
    servis_musteri.nlargest(10, "tutar")[
        [
            "kayit_id",
            "musteri_id",
            "arac_id",
            "servis_tarihi",
            "servis_tipi",
            "servis_merkezi",
            "tutar"
        ]
    ]
)

# Negatif tutarlı servis kayıtları
print("\n--- Negatif tutarlı servis kayıtları ---")
display(
    servis_musteri.loc[
        servis_musteri["tutar"] < 0,
        [
            "kayit_id",
            "musteri_id",
            "arac_id",
            "servis_tarihi",
            "servis_tipi",
            "tutar"
        ]
    ].sort_values("tutar")
)

# En yüksek toplam harcamaya sahip 10 müşteri
print("\n--- En yüksek toplam harcamaya sahip 10 müşteri ---")
display(
    rfm.nlargest(10, "monetary")[
        ["musteri_id", "frequency", "monetary", "recency"]
    ]
)
```

```python
# Kesim tarihi itibarıyla bilinebilecek araçlar
analiz_araclari = araclar[
    (araclar["musteri_id"].isin(analiz_musterileri["musteri_id"])) &
    (araclar["satin_alma_tarihi"] <= KESIM_TARIHI)
].copy()

# Araç özelliklerini müşteri seviyesinde özetle
arac_ozet = (
    analiz_araclari
    .groupby("musteri_id")
    .agg(
        arac_sayisi=("arac_id", "nunique"),
        ortalama_model_yili=("model_yili", "mean"),
        ortalama_ilk_km=("ilk_km", "mean"),
        aktif_garantili_arac_sayisi=(
            "garanti_bitis_tarihi",
            lambda x: (x >= KESIM_TARIHI).sum()
        )
    )
    .reset_index()
)

# Araç bilgileri bulunmayan müşteriler için araç sayısı 0
print("--- Araç sahipliği özeti ---")
print("Kesim tarihine kadar satın alınmış araç sayısı:", len(analiz_araclari))
print("Araç bilgisi bulunan müşteri sayısı:", arac_ozet["musteri_id"].nunique())
print("Analiz evrenindeki toplam müşteri sayısı:", len(analiz_musterileri))

print("\n--- Müşteri başına araç sayısı dağılımı ---")
display(
    arac_ozet["arac_sayisi"]
    .value_counts()
    .sort_index()
    .rename_axis("arac_sayisi")
    .to_frame("musteri_sayisi")
)

print("\n--- Araç özellikleri ---")
display(
    arac_ozet[
        [
            "arac_sayisi",
            "ortalama_model_yili",
            "ortalama_ilk_km",
            "aktif_garantili_arac_sayisi"
        ]
    ].describe().round(2)
)

print("\n--- Araç sayısı en yüksek 10 müşteri ---")
display(
    arac_ozet
    .sort_values("arac_sayisi", ascending=False)
    .head(10)
)
```

```python
# Kesim tarihine kadar olan geri bildirimler
geri_bildirim_gozlem = geri_bildirimler[
    geri_bildirimler["tarih"] <= KESIM_TARIHI
].copy()

print("--- Geri bildirim veri kontrolü ---")
print("Toplam geri bildirim:", len(geri_bildirimler))
print("Gözlem dönemindeki geri bildirim:", len(geri_bildirim_gozlem))

print("\n--- Geri bildirim sütunları ---")
print(geri_bildirim_gozlem.dtypes)

print("\n--- Geri bildirim eksik değerleri ---")
print(geri_bildirim_gozlem.isna().sum())

print("\n--- Geri bildirim tarih aralığı ---")
print("İlk tarih:", geri_bildirim_gozlem["tarih"].min())
print("Son tarih:", geri_bildirim_gozlem["tarih"].max())

print("\n--- Geri bildirim örnekleri ---")
display(
    geri_bildirim_gozlem[
        ["geri_bildirim_id", "arac_id", "tarih", "kanal", "metin"]
    ].head(10)
)
```

```python
# Geri bildirimleri araç tablosu üzerinden müşteriyle eşleştiRME
geri_bildirim_musteri = geri_bildirim_gozlem.merge(
    araclar[["arac_id", "musteri_id"]],
    on="arac_id",
    how="left",
    validate="many_to_one",
    indicator=True
)

print("--- Müşteri eşleştirme kontrolü ---")
print("Toplam geri bildirim:", len(geri_bildirim_musteri))
print(
    "Eşleşen geri bildirim:",
    (geri_bildirim_musteri["_merge"] == "both").sum()
)
print(
    "Eşleşmeyen geri bildirim:",
    (geri_bildirim_musteri["_merge"] == "left_only").sum()
)

print(
    "\nEşleşen farklı müşteri sayısı:",
    geri_bildirim_musteri["musteri_id"].nunique()
)

print(
    "Eksik müşteri ID sayısı:",
    geri_bildirim_musteri["musteri_id"].isna().sum()
)
```

```python
# Müşteri seviyesinde geri bildirim özeti
geri_bildirim_ozet = (
    geri_bildirim_musteri
    .groupby("musteri_id")
    .agg(
        geri_bildirim_sayisi=("geri_bildirim_id", "nunique"),
        metni_olan_geri_bildirim_sayisi=("metin", "count"),
        farkli_kanal_sayisi=("kanal", "nunique"),
        ilk_geri_bildirim_tarihi=("tarih", "min"),
        son_geri_bildirim_tarihi=("tarih", "max")
    )
    .reset_index()
)

print("--- Geri bildirim özeti ---")
print("Özet satır sayısı:", len(geri_bildirim_ozet))
print(
    "Tekil müşteri sayısı:",
    geri_bildirim_ozet["musteri_id"].nunique()
)

print("\n--- Geri bildirim sayısı dağılımı ---")
display(
    geri_bildirim_ozet["geri_bildirim_sayisi"]
    .value_counts()
    .sort_index()
    .rename_axis("Geri bildirim sayısı")
    .to_frame("Müşteri sayısı")
)

print("\n--- Özet istatistikler ---")
display(
    geri_bildirim_ozet[
        [
            "geri_bildirim_sayisi",
            "metni_olan_geri_bildirim_sayisi",
            "farkli_kanal_sayisi"
        ]
    ].describe()
)

print("\n--- İlk 10 müşteri ---")
display(geri_bildirim_ozet.head(10))
```

```python
# Müşteri temel bilgileri
musteri_ozet = analiz_musterileri[
    [
        "musteri_id",
        "sehir",
        "musteri_tipi",
        "kayit_kanali"
    ]
].copy()

# Ana müşteri analiz tablosunu oluştur
musteri_analiz = (
    musteri_ozet
    .merge(
        rfm,
        on="musteri_id",
        how="left",
        validate="one_to_one"
    )
    .merge(
        arac_ozet[
            [
                "musteri_id",
                "arac_sayisi",
                "ortalama_model_yili",
                "aktif_garantili_arac_sayisi"
            ]
        ],
        on="musteri_id",
        how="left",
        validate="one_to_one"
    )
    .merge(
        geri_bildirim_ozet,
        on="musteri_id",
        how="left",
        validate="one_to_one"
    )
)

# Geri bildirim bırakmayan müşteriler için sayısal adetleri 0 
adet_kolonlari = [
    "geri_bildirim_sayisi",
    "metni_olan_geri_bildirim_sayisi",
    "farkli_kanal_sayisi"
]

musteri_analiz[adet_kolonlari] = (
    musteri_analiz[adet_kolonlari]
    .fillna(0)
    .astype(int)
)

print("--- Ana müşteri analiz tablosu ---")
print("Satır sayısı:", len(musteri_analiz))
print("Sütun sayısı:", len(musteri_analiz.columns))
print("Tekil müşteri ID:", musteri_analiz["musteri_id"].is_unique)

print("\n--- Sütunlar ---")
print(musteri_analiz.columns.tolist())

print("\n--- Eksik değerler ---")
display(
    musteri_analiz.isna().sum().to_frame("Eksik değer sayısı")
)

print("\n--- İlk 5 müşteri ---")
display(musteri_analiz.head())
```

```python
print("--- Ana tablo kalite kontrolleri ---")

# 1. Müşteri sayısı
print(
    "Ana tablo müşteri sayısı doğru mu?",
    len(musteri_analiz) == len(analiz_musterileri)
)

# 2. Analiz popülasyonundaki geri bildirim kayıtlarını seç
populasyon_ici_geri_bildirimler = geri_bildirim_musteri[
    geri_bildirim_musteri["musteri_id"].isin(
        analiz_musterileri["musteri_id"]
    )
]

# 3. Geri bildirim sayısı
print(
    "Geri bildirim sayısı toplamı doğru mu?",
    musteri_analiz["geri_bildirim_sayisi"].sum()
    == len(populasyon_ici_geri_bildirimler)
)

# 4. Geri bildirim bırakan müşteri sayısı
print(
    "Geri bildirimli müşteri sayısı doğru mu?",
    (musteri_analiz["geri_bildirim_sayisi"] > 0).sum()
    == populasyon_ici_geri_bildirimler["musteri_id"].nunique()
)

# 5. Tarih kesim kontrolü
tarih_kolonlari = [
    "son_servis_tarihi",
    "ilk_geri_bildirim_tarihi",
    "son_geri_bildirim_tarihi"
]

for kolon in tarih_kolonlari:
    print(
        f"{kolon} kesim tarihini aşıyor mu?",
        (musteri_analiz[kolon] > KESIM_TARIHI).any()
    )

# 6. Araç sayısı kontrolü
print(
    "Araç sayısı toplamı doğru mu?",
    musteri_analiz["arac_sayisi"].sum()
    == len(analiz_araclari)
)

# 7. Son tablo
print("\nAna tablo boyutu:", musteri_analiz.shape)
```

```python
# Geri bildirimlerin analiz popülasyonuyla tutarlılık kontrolü
geri_bildirim_musteri_id = set(
    geri_bildirim_ozet["musteri_id"]
)

analiz_musteri_id = set(
    analiz_musterileri["musteri_id"]
)

# Geri bildirim bırakmış ancak analiz popülasyonunda olmayan müşteriler
populasyon_disi_musteriler = (
    geri_bildirim_musteri_id - analiz_musteri_id
)

# Analiz popülasyonunda bulunan geri bildirimli müşteriler
populasyon_ici_musteriler = (
    geri_bildirim_musteri_id & analiz_musteri_id
)

print("Geri bildirim bırakan toplam müşteri:",
      len(geri_bildirim_musteri_id))

print("Analiz popülasyonundaki geri bildirimli müşteri:",
      len(populasyon_ici_musteriler))

print("Analiz popülasyonu dışındaki geri bildirimli müşteri:",
      len(populasyon_disi_musteriler))

# Popülasyon dışındaki müşterilerin geri bildirim sayısı
disaridaki_geri_bildirimler = geri_bildirim_musteri[
    geri_bildirim_musteri["musteri_id"].isin(
        populasyon_disi_musteriler
    )
]

print("\nPopülasyon dışındaki geri bildirim kayıt sayısı:",
      len(disaridaki_geri_bildirimler))

print("\nPopülasyon dışındaki müşteri başına geri bildirim:")
display(
    disaridaki_geri_bildirimler
    .groupby("musteri_id")
    .size()
    .value_counts()
    .sort_index()
    .rename_axis("Geri bildirim sayısı")
    .to_frame("Müşteri sayısı")
)
```

```python
# Gözlem dönemindeki servis kayıtlarının memnuniyet analizi
print("--- Genel memnuniyet kontrolü ---")

print("Toplam servis kaydı:", len(servis_gozlem))

print(
    "Eksik memnuniyet puanı:",
    servis_gozlem["memnuniyet_puani"].isna().sum()
)

print(
    "Eksik memnuniyet oranı (%):",
    round(
        servis_gozlem["memnuniyet_puani"].isna().mean() * 100,
        2
    )
)

print("\n--- Memnuniyet puanı dağılımı ---")

display(
    servis_gozlem["memnuniyet_puani"]
    .value_counts(dropna=False)
    .sort_index()
    .to_frame("Kayıt sayısı")
)

print("\n--- Memnuniyet istatistikleri ---")

display(
    servis_gozlem["memnuniyet_puani"].describe()
)
```

```python
# Servis merkezlerine göre memnuniyet verisinin incelenmesi

merkez_memnuniyet_kontrol = (
    servis_gozlem
    .groupby("servis_merkezi")
    .agg(
        toplam_servis=("kayit_id", "count"),
        puanli_kayit=("memnuniyet_puani", "count"),
        ortalama_memnuniyet=("memnuniyet_puani", "mean"),
        medyan_memnuniyet=("memnuniyet_puani", "median")
    )
    .reset_index()
)

merkez_memnuniyet_kontrol["eksik_puan_sayisi"] = (
    merkez_memnuniyet_kontrol["toplam_servis"]
    - merkez_memnuniyet_kontrol["puanli_kayit"]
)

merkez_memnuniyet_kontrol["eksik_puan_orani (%)"] = (
    merkez_memnuniyet_kontrol["eksik_puan_sayisi"]
    / merkez_memnuniyet_kontrol["toplam_servis"]
    * 100
).round(2)

merkez_memnuniyet_kontrol[
    "ortalama_memnuniyet"
] = merkez_memnuniyet_kontrol[
    "ortalama_memnuniyet"
].round(2)

merkez_memnuniyet_kontrol[
    "medyan_memnuniyet"
] = merkez_memnuniyet_kontrol[
    "medyan_memnuniyet"
].round(2)

display(
    merkez_memnuniyet_kontrol.sort_values(
        "toplam_servis",
        ascending=False
    )
)
```

```python
from scipy.stats import kruskal

# Her servis merkezi için eksik olmayan memnuniyet puanları
memnuniyet_gruplari = [
    grup["memnuniyet_puani"].dropna()
    for _, grup in servis_gozlem.groupby("servis_merkezi")
]

# Kruskal-Wallis testi
kruskal_istatistigi, p_degeri = kruskal(*memnuniyet_gruplari)

print("--- Kruskal-Wallis Testi ---")
print("Test istatistiği:", round(kruskal_istatistigi, 4))
print("P-değeri:", p_degeri)

if p_degeri < 0.05:
    print(
        "\nSonuç: Servis merkezlerinin memnuniyet puanı "
        "dağılımları arasında istatistiksel olarak anlamlı fark var."
    )
else:
    print(
        "\nSonuç: Servis merkezlerinin memnuniyet puanı "
        "dağılımları arasında istatistiksel olarak anlamlı fark "
        "olduğuna dair yeterli kanıt yok."
    )
```

```python
from itertools import combinations
from scipy.stats import mannwhitneyu

# Her merkezin eksik olmayan memnuniyet puanları
merkez_gruplari = {
    merkez: grup["memnuniyet_puani"].dropna()
    for merkez, grup in servis_gozlem.groupby("servis_merkezi")
}

# Tüm merkez çiftlerini karşılaştır
ikili_sonuclar = []

for (merkez_1, puanlar_1), (merkez_2, puanlar_2) in combinations(
    merkez_gruplari.items(), 2
):
    istatistik, p_degeri = mannwhitneyu(
        puanlar_1,
        puanlar_2,
        alternative="two-sided"
    )

    ikili_sonuclar.append({
        "merkez_1": merkez_1,
        "merkez_2": merkez_2,
        "n_1": len(puanlar_1),
        "n_2": len(puanlar_2),
        "u_istatistigi": istatistik,
        "p_degeri": p_degeri
    })

ikili_karsilastirma = pd.DataFrame(ikili_sonuclar)

# Bonferroni düzeltmesi
karsilastirma_sayisi = len(ikili_karsilastirma)

ikili_karsilastirma["p_duzeltilmis"] = (
    ikili_karsilastirma["p_degeri"] * karsilastirma_sayisi
).clip(upper=1)

ikili_karsilastirma["anlamli_mi"] = (
    ikili_karsilastirma["p_duzeltilmis"] < 0.05
)

display(
    ikili_karsilastirma.sort_values("p_duzeltilmis")
)
```

```python
# Anlamlı bulunan SM-Ankara1 - SM-Maslak karşılaştırmasının etki büyüklüğü

anlamli_sonuc = ikili_karsilastirma[
    (ikili_karsilastirma["merkez_1"] == "SM-Ankara1") &
    (ikili_karsilastirma["merkez_2"] == "SM-Maslak")
].iloc[0]

n1 = anlamli_sonuc["n_1"]
n2 = anlamli_sonuc["n_2"]
u = anlamli_sonuc["u_istatistigi"]

# Rank-biserial correlation
rank_biserial = (2 * u / (n1 * n2)) - 1

print("SM-Ankara1 - SM-Maslak")
print("Rank-biserial etki büyüklüğü:", round(rank_biserial, 4))
print("Mutlak etki büyüklüğü:", round(abs(rank_biserial), 4))
```

**Değerlendirme:** Servis merkezlerinin memnuniyet puanı dağılımları arasında istatistiksel olarak anlamlı fark bulunmuştur (Kruskal-Wallis, p=0,0021). Bonferroni düzeltmeli ikili karşılaştırmalarda yalnızca SM-Ankara1 ile SM-Maslak arasındaki fark anlamlıdır (p=0,0198). Ancak etki büyüklüğünün düşük olması (0,0418), bu farkın pratik açıdan sınırlı olduğunu göstermektedir.

```python
# Hedef dönemindeki servis kayıtlarını müşteri bazında özetleme

hedef_musteri_ozet = (
    hedef_servis_musteri
    .dropna(subset=["musteri_id"])
    .groupby("musteri_id")
    .agg(
        hedef_donem_servis_sayisi=("kayit_id", "nunique")
    )
    .reset_index()
)

# Hedef dönem servis özetini analiz tablosuna ekleme
musteri_analiz = musteri_analiz.merge(
    hedef_musteri_ozet,
    on="musteri_id",
    how="left",
    validate="one_to_one"
)

# Hedef dönemde servise gelmeyen müşterilerin sayısını 0 yapma
musteri_analiz["hedef_donem_servis_sayisi"] = (
    musteri_analiz["hedef_donem_servis_sayisi"]
    .fillna(0)
    .astype(int)
)

# Churn hedef değişkenini oluşturma
# 0: Hedef dönemde en az bir servis kaydı var
# 1: Hedef dönemde hiç servis kaydı yok
musteri_analiz["churn"] = (
    musteri_analiz["hedef_donem_servis_sayisi"] == 0
).astype(int)

print("--- Churn hedef değişkeni kontrolü ---")
print("Analiz popülasyonu:", len(musteri_analiz))
print("Tekil müşteri sayısı:", musteri_analiz["musteri_id"].nunique())

print("\nChurn dağılımı:")
print(
    musteri_analiz["churn"]
    .value_counts()
    .sort_index()
    .rename(index={0: "Churn yok", 1: "Churn var"})
)

print("\nChurn oranları (%):")
print(
    (musteri_analiz["churn"].value_counts(normalize=True) * 100)
    .sort_index()
    .rename(index={0: "Churn yok", 1: "Churn var"})
    .round(2)
)

print("\nHedef dönemindeki servis sayısı kontrolü:")
print(musteri_analiz["hedef_donem_servis_sayisi"].describe())

print("\nEksik churn etiketi:", musteri_analiz["churn"].isna().sum())
print(
    "Müşteri ID benzersiz mi?",
    musteri_analiz["musteri_id"].is_unique
)
```
