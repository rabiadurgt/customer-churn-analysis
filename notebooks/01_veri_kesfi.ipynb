Tarih alanları incelendiğinde müşteri, araç ve geri bildirim tablolarındaki tarih sütunlarının tutarlı biçimde YYYY-MM-DD formatında saklandığı görülmüştür. Servis kayıtlarında ise dört farklı tarih formatı tespit edilmiştir. Bu nedenle analiz öncesinde tarih alanlarının standart bir tarih veri tipine dönüştürülmesi gerekmektedir.

```python
# Tarih dönüşümleri: Ham sütunları koruyarak yeni sütunlar oluşturuyoruz


musteriler["ilk_temas_tarihi_dt"] = pd.to_datetime(
    musteriler["ilk_temas_tarihi"],
    format="%Y-%m-%d",
    errors="coerce"
)

araclar["satin_alma_tarihi_dt"] = pd.to_datetime(
    araclar["satin_alma_tarihi"],
    format="%Y-%m-%d",
    errors="coerce"
)

araclar["garanti_bitis_tarihi_dt"] = pd.to_datetime(
    araclar["garanti_bitis_tarihi"],
    format="%Y-%m-%d",
    errors="coerce"
)

geri_bildirimler["tarih_dt"] = pd.to_datetime(
    geri_bildirimler["tarih"],
    format="%Y-%m-%d",
    errors="coerce"
)

# Servis tarihlerini formatlarına göre ayrı ayrı dönüştürme
tarih_metni = servis_kayitlari["servis_tarihi"].astype(str).str.strip()

servis_tarihi_dt = pd.Series(
    pd.NaT,
    index=servis_kayitlari.index,
    dtype="datetime64[ns]"
)

# 1. ISO: YYYY-MM-DD
maske_iso = tarih_metni.str.fullmatch(r"\d{4}-\d{2}-\d{2}")

servis_tarihi_dt.loc[maske_iso] = pd.to_datetime(
    tarih_metni.loc[maske_iso],
    format="%Y-%m-%d",
    errors="coerce"
)

# 2. ISO datetime: YYYY-MM-DD HH:MM:SS
maske_iso_datetime = tarih_metni.str.fullmatch(
    r"\d{4}-\d{2}-\d{2} \d{2}:\d{2}:\d{2}"
)

servis_tarihi_dt.loc[maske_iso_datetime] = pd.to_datetime(
    tarih_metni.loc[maske_iso_datetime],
    format="%Y-%m-%d %H:%M:%S",
    errors="coerce"
)

# 3. DD/MM/YYYY
maske_slash = tarih_metni.str.fullmatch(r"\d{2}/\d{2}/\d{4}")

servis_tarihi_dt.loc[maske_slash] = pd.to_datetime(
    tarih_metni.loc[maske_slash],
    format="%d/%m/%Y",
    errors="coerce"
)

# 4. DD.MM.YYYY
maske_dot = tarih_metni.str.fullmatch(r"\d{2}\.\d{2}\.\d{4}")

servis_tarihi_dt.loc[maske_dot] = pd.to_datetime(
    tarih_metni.loc[maske_dot],
    format="%d.%m.%Y",
    errors="coerce"
)

servis_kayitlari["servis_tarihi_dt"] = servis_tarihi_dt

print("Toplam kayıt:", len(servis_kayitlari))
print("Dönüştürülemeyen tarih:", servis_kayitlari["servis_tarihi_dt"].isna().sum())


print("Tarih sütunları oluşturuldu.")
```

```python

GOZLEM_BASLANGIC = pd.Timestamp("2022-01-01")
GOZLEM_BITIS = pd.Timestamp("2025-06-30")

HEDEF_BASLANGIC = pd.Timestamp("2025-07-01")
HEDEF_BITIS = pd.Timestamp("2026-06-30")

servis_tarihi = servis_kayitlari["servis_tarihi_dt"]

# Zaman pencerelerine göre kayıtları sınıflandırma

donem_kontrolu = {
    "Gözlem dönemi öncesi": servis_tarihi < GOZLEM_BASLANGIC,

    "Gözlem dönemi": (
        (servis_tarihi >= GOZLEM_BASLANGIC) &
        (servis_tarihi <= GOZLEM_BITIS)
    ),

    "Hedef dönemi": (
        (servis_tarihi >= HEDEF_BASLANGIC) &
        (servis_tarihi <= HEDEF_BITIS)
    ),

    "Hedef dönemi sonrası": servis_tarihi > HEDEF_BITIS
}

for donem, maske in donem_kontrolu.items():
    print(f"{donem}: {maske.sum():,} kayıt")

print("\nToplam servis kaydı:", len(servis_kayitlari))
print(
    "Sınıflandırılan toplam kayıt:",
    sum(maske.sum() for maske in donem_kontrolu.values())
)
```

```python
# Veri profili özetini oluşturma

profil_kayitlari = []

for tablo_adi, df in veri_setleri.items():
    profil_kayitlari.append({
        "tablo": tablo_adi,
        "satir_sayisi": len(df),
        "kolon_sayisi": len(beklenen_kolonlar[tablo_adi]),
        "gercek_kolon_sayisi": len([
            kolon for kolon in df.columns
            if not kolon.endswith("_dt")
        ]),
        "eksik_hucre_sayisi": int(df.isna().sum().sum()),
        "tam_tekerrur_satir_sayisi": int(df.duplicated().sum()),
        "benzersiz_kayit_sayisi": (
            int(df.iloc[:, 0].nunique())
        )
    })

veri_profili = pd.DataFrame(profil_kayitlari)

ANALIZ_DIR = PROJECT_ROOT / "ciktilar" / "analiz"
ANALIZ_DIR.mkdir(parents=True, exist_ok=True)

profil_dosyasi = ANALIZ_DIR / "01_veri_profili.csv"
veri_profili.to_csv(profil_dosyasi, index=False, encoding="utf-8-sig")

print("Veri profili oluşturuldu:", profil_dosyasi)
display(veri_profili)
```
